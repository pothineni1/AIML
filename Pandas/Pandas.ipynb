## How to convert to Camel Case

Camel case (specifically, PascalCase) refers to a naming convention where the first letter of each word in a compound word is capitalized, with no intervening spaces or punctuation. For example, `hello world` becomes `HelloWorld`.

We'll define a function to achieve this and then apply it to the 'Name' column.

```python
def to_camel_case(name):
    # Split the name by spaces and convert each part to title case (first letter capitalized)
    camel_name = ''.join(word.capitalize() for word in name.split())
    return camel_name

# Apply the function to the 'Name' column and display the result
df2['Name_CamelCase'] = df2['Name'].apply(to_camel_case)
display(df2[['Name', 'Name_CamelCase']].head())
```
