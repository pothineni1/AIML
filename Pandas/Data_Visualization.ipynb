```python
import matplotlib.pyplot as plt
```

## Line Plot

```python
plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
sales = [100,150,200,180,190,250,300,280,320,350,400,487]
```

```python
from matplotlib.lines import lineStyles
plt.plot(months, sales, color = 'red', linestyle='--',marker='o',linewidth=2,markersize=5)
plt.title("Sales Per Month")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
import numpy as np
```

```python
arr2 = np.array(sales)
profit = arr2 *0.4
profit = profit.tolist()
profit
```

```python
from matplotlib.lines import lineStyles
plt.plot(months, sales, marker='o',label='Sales')
plt.plot(months, profit, marker='o',label='Profit')
plt.title("Monthly sales and profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.show()
```

```python
profit2 = arr2 *0.6
```

```python
plt.plot(months, sales, marker='o',label='Sales')
plt.plot(months, profit, marker='o',label='Profit')
plt.plot(months, profit2, marker='o',label='Profit2')
plt.title("Monthly sales and profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
import plotly.express as px
```

```python
px.line(y=sales, x=months, title="Sales Per Month")
```

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/pothineni1/AIML/refs/heads/main/SuperStore.csv')
df
```

```python
df.columns
```

```python
df.info()
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date']=pd.to_datetime(df['Ship Date'], dayfirst=True)
df.info()
```

## Sales BY category -- Bar chart

```python
df['Category'].unique()
```

```python
category_sales = df.groupby('Category')['Sales'].sum().reset_index()
category_sales
```

```python
import plotly.express as px
```

```python
category_sales_region = df.groupby(['Category','Region'])['Sales'].sum().reset_index()
category_sales_region
```

```python
px.bar(category_sales_region, x='Category', y='Sales', color='Region', barmode='group', title = 'Sales by Category and Region')
```

```python
px.line(df,x = 'Order Date', y ='Sales', title='Sales over time')
```
