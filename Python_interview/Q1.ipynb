```python
req_skills = {'Pyhton','aws','snowflake','docker','sql'}
type(req_skills)
```

```python
table_data = { "Candidate_id": ['C001','C002','C003','C004','C005'],
              "skills": ["Pyhton,aws,docker,sql",
                         "Pyhton,aws,snowflake,docker,sql",
                         "java,sql",
                         "Pyhton,aws,docker,snowflake",
                         "Pyhton,docker,sql"]
              }
table_data
```

```python
import pandas as pd
```

```python
df_table = pd.DataFrame(table_data)
df_table
```

```python
def percentage_match(skills):
  cand_skilss = {s.strip() for s in skills.split(",")}
  matched_skills = cand_skilss.intersection(req_skills)

  return  (len(matched_skills)/len(req_skills))*100
```

```python
df_table["percentage_match"] = df_table['skills'].apply(percentage_match)
df_table
```

```python
result = df_table[df_table["percentage_match"] >= 80]
display(result)
```
